```r
# 加载必要包
library(dplyr)
library(zoo)

########################################
# 1. 处理 crime_day.csv
########################################

# 读取白天数据
crime_day <- read.csv("crime_day.csv")

# (A) 先按 (name, YEAR, MONTH) 统计各 division 案件数
crime_day_monthly <- crime_day %>%
  group_by(geo_local_area , YEAR, MONTH) %>%
  summarise(
    crime_count = n(),
    .groups = "drop"
  ) %>%
  mutate(date = as.yearmon(paste(YEAR, MONTH), "%Y %m")) %>%
  arrange(geo_local_area , date) %>%
  group_by(geo_local_area ) %>%
  mutate(
    prev_crime_count = lag(crime_count),
    pct_change = 100 * (crime_count - prev_crime_count) / prev_crime_count
  ) %>%
  ungroup()

# 写出各 division 白天环比变化
write.csv(crime_day_monthly, "crime_day_monthly_pct_change.csv", row.names = FALSE)

# (B) 再计算**全温哥华**：按 (YEAR, MONTH) 汇总
crime_day_monthly_van <- crime_day %>%
  group_by(YEAR, MONTH) %>%
  summarise(
    crime_count = n(),
    .groups = "drop"
  ) %>%
  mutate(date = as.yearmon(paste(YEAR, MONTH), "%Y %m")) %>%
  arrange(date) %>%
  mutate(
    prev_crime_count = lag(crime_count),
    pct_change = 100 * (crime_count - prev_crime_count) / prev_crime_count
  )

# 写出全温哥华 白天环比变化
write.csv(crime_day_monthly_van, "crime_day_monthly_pct_change_van.csv", row.names = FALSE)

########################################
# 2. 处理 crime_night.csv
########################################

# 读取夜晚数据
crime_night <- read.csv("crime_night.csv")

# (A) 各 division 夜间案件
crime_night_monthly <- crime_night %>%
  group_by(geo_local_area , YEAR, MONTH) %>%
  summarise(
    crime_count = n(),
    .groups = "drop"
  ) %>%
  mutate(date = as.yearmon(paste(YEAR, MONTH), "%Y %m")) %>%
  arrange(geo_local_area , date) %>%
  group_by(geo_local_area ) %>%
  mutate(
    prev_crime_count = lag(crime_count),
    pct_change = 100 * (crime_count - prev_crime_count) / prev_crime_count
  ) %>%
  ungroup()

# 写出各 division 夜晚环比变化
write.csv(crime_night_monthly, "crime_night_monthly_pct_change.csv", row.names = FALSE)

# (B) 全温哥华 夜间案件
crime_night_monthly_van <- crime_night %>%
  group_by(YEAR, MONTH) %>%
  summarise(
    crime_count = n(),
    .groups = "drop"
  ) %>%
  mutate(date = as.yearmon(paste(YEAR, MONTH), "%Y %m")) %>%
  arrange(date) %>%
  mutate(
    prev_crime_count = lag(crime_count),
    pct_change = 100 * (crime_count - prev_crime_count) / prev_crime_count
  )

# 写出全温哥华 夜晚环比变化
write.csv(crime_night_monthly_van, "crime_night_monthly_pct_change_van.csv", row.names = FALSE)

cat("处理完成：\n",
    "1) crime_day_monthly_pct_change.csv\n",
    "2) crime_day_monthly_pct_change_van.csv\n",
    "3) crime_night_monthly_pct_change.csv\n",
    "4) crime_night_monthly_pct_change_van.csv\n")
```
