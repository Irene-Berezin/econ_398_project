```python
# Analyzing data in neighborhood_data with respect to date and crime_count

# Check for missing values in crime_count
print(f"Missing values in crime_count: {neighborhood_data['crime_count'].isna().sum()}")
print(f"Total rows in dataset: {len(neighborhood_data)}")
print(f"Percentage of missing crime_count values: {neighborhood_data['crime_count'].isna().sum() / len(neighborhood_data) * 100:.2f}%")

# Examine the date column
print("\nDate information:")
# Convert date to datetime if it's not already
if pd.api.types.is_datetime64_any_dtype(neighborhood_data['date']):
    date_column = 'date'
else:
    # Try to find the date column or convert if necessary
    date_candidates = [col for col in neighborhood_data.columns if 'date' in col.lower()]
    if date_candidates:
        date_column = date_candidates[0]
        neighborhood_data[date_column] = pd.to_datetime(neighborhood_data[date_column], errors='coerce')
    else:
        print("No date column found. Please specify the correct date column.")

# If we have a valid date column, analyze it
if 'date_column' in locals():
    print(f"Date range: {neighborhood_data[date_column].min()} to {neighborhood_data[date_column].max()}")
    print(f"Number of unique dates: {neighborhood_data[date_column].nunique()}")
    
    # Group by date and count observations
    date_counts = neighborhood_data.groupby(date_column).size()
    print(f"\nNumber of observations per date (first 5):\n{date_counts.head()}")
    
    # Analyze crime_count by date
    crime_by_date = neighborhood_data.groupby(date_column)['crime_count'].agg(['mean', 'min', 'max', 'count'])
    print(f"\nCrime count statistics by date (first 5 dates):\n{crime_by_date.head()}")
    
    # Plot the data
    import matplotlib.pyplot as plt
    import seaborn as sns
    
    plt.figure(figsize=(12, 6))
    
    # Create subplot for time series
    plt.subplot(1, 2, 1)
    sns.lineplot(x=date_column, y='crime_count', data=neighborhood_data)
    plt.title('Crime Count Over Time')
    plt.xticks(rotation=45)
    
    # Create subplot for distribution
    plt.subplot(1, 2, 2)
    sns.histplot(data=neighborhood_data, x='crime_count', kde=True)
    plt.title('Distribution of Crime Count Values')
    
    plt.tight_layout()
    plt.show()
    
    # Optional: Check for seasonal patterns if data spans multiple years
    if neighborhood_data[date_column].max().year - neighborhood_data[date_column].min().year > 0:
        neighborhood_data['month'] = neighborhood_data[date_column].dt.month
        monthly_avg = neighborhood_data.groupby('month')['crime_count'].mean()
        
        plt.figure(figsize=(10, 5))
        sns.barplot(x=monthly_avg.index, y=monthly_avg.values)
        plt.title('Average Crime Count by Month')
        plt.xlabel('Month')
        plt.ylabel('Average Crime Count')
        plt.show()
```

```python
# Analyzing neighborhood data distribution and patterns - fixed version

# Check how many neighborhoods are in the dataset
print(f"Number of unique neighborhoods: {neighborhood_data['neighborhood'].nunique()}")
print(f"List of neighborhoods: {sorted(neighborhood_data['neighborhood'].unique())}")

# Count observations per neighborhood
neighborhood_counts = neighborhood_data['neighborhood'].value_counts().sort_values(ascending=False)
print(f"\nObservations per neighborhood (top 10):\n{neighborhood_counts.head(10)}")
print(f"\nObservations per neighborhood (bottom 10):\n{neighborhood_counts.tail(10)}")

# Check for data completeness across neighborhoods
neighborhood_date_counts = neighborhood_data.groupby('neighborhood')[date_column].nunique()
print(f"\nNumber of unique dates per neighborhood (sample):")
print(f"Max dates: {neighborhood_date_counts.max()}")
print(f"Min dates: {neighborhood_date_counts.min()}")
print(f"Mean dates: {neighborhood_date_counts.mean():.1f}")
print(f"\nNeighborhoods with fewest dates:\n{neighborhood_date_counts.sort_values().head()}")

# Analyze crime_count by neighborhood
crime_by_neighborhood = neighborhood_data.groupby('neighborhood')['crime_count'].agg(['mean', 'std', 'min', 'max', 'count'])
print(f"\nCrime count statistics by neighborhood (sample):")
print(f"Highest average crime count:\n{crime_by_neighborhood.sort_values('mean', ascending=False).head(5)}")
print(f"Lowest average crime count:\n{crime_by_neighborhood.sort_values('mean').head(5)}")

# Visualization of crime_count by neighborhood
plt.figure(figsize=(14, 8))

# Boxplot of crime_count by neighborhood
plt.subplot(2, 1, 1)
# If there are many neighborhoods, limit to top 15 by mean value for readability
if neighborhood_data['neighborhood'].nunique() > 15:
    top_neighborhoods = crime_by_neighborhood.sort_values('mean', ascending=False).head(15).index
    plot_data = neighborhood_data[neighborhood_data['neighborhood'].isin(top_neighborhoods)]
    sns.boxplot(x='neighborhood', y='crime_count', data=plot_data)
    plt.title('Distribution of Crime Count Values by Top 15 Neighborhoods')
else:
    sns.boxplot(x='neighborhood', y='crime_count', data=neighborhood_data)
    plt.title('Distribution of Crime Count Values by Neighborhood')
plt.xticks(rotation=90)
plt.tight_layout()

# Time series for top 5 neighborhoods by mean crime count value
plt.subplot(2, 1, 2)
top_5_neighborhoods = crime_by_neighborhood.sort_values('mean', ascending=False).head(5).index
for hood in top_5_neighborhoods:
    hood_data = neighborhood_data[neighborhood_data['neighborhood'] == hood]
    sns.lineplot(x=date_column, y='crime_count', data=hood_data, label=hood)
plt.title('Crime Count Over Time for Top 5 Neighborhoods')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Comprehensive analysis of neighborhood data with focus on crime count

# Calculate summary statistics of crime_count by neighborhood
crime_count_stats = neighborhood_data.groupby('neighborhood')['crime_count'].agg([
    'count', 'mean', 'median', 'min', 'max', 'std'
]).reset_index()

# Sort by mean crime_count descending
crime_count_stats = crime_count_stats.sort_values('mean', descending=False)

print("Crime count statistics by neighborhood:")
print(crime_count_stats)

# Visualization of crime count distribution across neighborhoods
plt.figure(figsize=(14, 10))

# Bar chart of average crime count per neighborhood
plt.subplot(2, 1, 1)
sns.barplot(x='neighborhood', y='mean', data=crime_count_stats)
plt.title('Average Crime Count per Neighborhood')
plt.xticks(rotation=90)
plt.ylabel('Mean Crime Count')
plt.xlabel('')

# Scatter plot of crime count variability (mean vs std)
plt.subplot(2, 1, 2)
sns.scatterplot(x='mean', y='std', data=crime_count_stats, s=100)
# Add neighborhood labels to points
for i, row in crime_count_stats.iterrows():
    plt.text(row['mean'], row['std'], row['neighborhood'], fontsize=9)
plt.title('Crime Count Variability by Neighborhood')
plt.xlabel('Mean Crime Count')
plt.ylabel('Standard Deviation of Crime Count')

plt.tight_layout()
plt.show()

# Time series analysis of crime counts
plt.figure(figsize=(16, 8))

# Get overall trend for all neighborhoods combined
crime_by_date = neighborhood_data.groupby(date_column)['crime_count'].mean().reset_index()
crime_by_date.columns = [date_column, 'average_crime_count']

# Time series plot of average crime count over time
sns.lineplot(x=date_column, y='average_crime_count', data=crime_by_date)
plt.title('Average Crime Count Over Time (All Neighborhoods)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Statistical analysis of crime count patterns
print("\nOverall crime count statistics:")
print(f"Average crime count across all neighborhoods: {neighborhood_data['crime_count'].mean():.2f}")
print(f"Median crime count: {neighborhood_data['crime_count'].median():.2f}")
print(f"Standard deviation: {neighborhood_data['crime_count'].std():.2f}")
print(f"Maximum crime count: {neighborhood_data['crime_count'].max():.2f} (Neighborhood: {neighborhood_data.loc[neighborhood_data['crime_count'].idxmax(), 'neighborhood']})")
print(f"Minimum crime count: {neighborhood_data['crime_count'].min():.2f} (Neighborhood: {neighborhood_data.loc[neighborhood_data['crime_count'].idxmin(), 'neighborhood']})")

# Create a heatmap of crime count over time if possible
if neighborhood_data[date_column].nunique() <= 50 and neighborhood_data['neighborhood'].nunique() <= 30:
    # Create a pivot table for the heatmap
    pivot_data = neighborhood_data.pivot_table(
        index='neighborhood',
        columns=date_column,
        values='crime_count',
        aggfunc='mean'
    )
    
    plt.figure(figsize=(16, 10))
    sns.heatmap(pivot_data, cmap='YlOrRd', linewidths=0.5)
    plt.title('Heatmap of Crime Count by Neighborhood Over Time')
    plt.xticks(rotation=90)
    plt.tight_layout()
    plt.show()
else:
    print("\nHeatmap not generated due to too many unique dates or neighborhoods (would be unreadable).")
```

# Comparing Day and Night Characteristics

Let's create time categories (day/night) and compare the characteristics between them

# Detailed Comparison: Day vs. Night Crime Data

Comparing key metrics between day and night crime data groups including geographical, temporal, demographic, and socioeconomic characteristics.

```python
# Detailed comparison between day and night crime data
print("\n--- DETAILED COMPARISON: DAY VS NIGHT CRIME DATA ---\n")

# Use the existing neighborhood_data and split it into day and night
# First, check if neighborhood_data is already loaded
if 'neighborhood_data' not in locals() or not isinstance(neighborhood_data, pd.DataFrame):
    print("Loading neighborhood_data from file...")
    neighborhood_data = pd.read_csv('neighborhood2016_2021_panel.csv')

# Display first few rows and columns in neighborhood_data to confirm structure
print("Sample of neighborhood_data:")
display(neighborhood_data.head())
print("Columns in neighborhood_data:", neighborhood_data.columns.tolist())

# Create day and night subsets based on time_category
crime_day = neighborhood_data[neighborhood_data['time_category'] == 'day'].copy()
crime_night = neighborhood_data[neighborhood_data['time_category'] == 'night'].copy()

# Display the number of rows in each subset
print(f"\nDay crime data: {len(crime_day)} rows")
print(f"Night crime data: {len(crime_night)} rows")

# Create a comparison dataframe for the key metrics
comparison_metrics = pd.DataFrame({
    'Metric': [
        'Number of observations',
        'Unique neighborhoods',
        'Unique years',
        'Unique months',
        'Unique dates',
        'Mean population',
        'Mean area_sq_km',
        'Mean population_density',
        'Mean income',
        'Mean crime count',
        'Mean NTL',
        'Downtown observations (is_downtown=1)'
    ]
})

# Calculate metrics for day crimes
day_metrics = [
    len(crime_day),
    crime_day['neighborhood'].nunique() if 'neighborhood' in crime_day.columns else 'N/A',
    crime_day['year'].nunique() if 'year' in crime_day.columns else 'N/A',
    crime_day['month'].nunique() if 'month' in crime_day.columns else 'N/A',
    crime_day['date'].nunique() if 'date' in crime_day.columns else 'N/A',
    crime_day['population'].mean() if 'population' in crime_day.columns else 'N/A',
    crime_day['area_sq_km'].mean() if 'area_sq_km' in crime_day.columns else 'N/A',
    crime_day['population_density'].mean() if 'population_density' in crime_day.columns else 'N/A',
    crime_day['income'].mean() if 'income' in crime_day.columns else 'N/A',
    crime_day['crime_count'].mean() if 'crime_count' in crime_day.columns else 'N/A',
    crime_day['ntl_mean'].mean() if 'ntl_mean' in crime_day.columns else 'N/A',
    crime_day[crime_day['is_downtown'] == 1].shape[0] if 'is_downtown' in crime_day.columns else 'N/A'
]

# Calculate metrics for night crimes
night_metrics = [
    len(crime_night),
    crime_night['neighborhood'].nunique() if 'neighborhood' in crime_night.columns else 'N/A',
    crime_night['year'].nunique() if 'year' in crime_night.columns else 'N/A',
    crime_night['month'].nunique() if 'month' in crime_night.columns else 'N/A',
    crime_night['date'].nunique() if 'date' in crime_night.columns else 'N/A',
    crime_night['population'].mean() if 'population' in crime_night.columns else 'N/A',
    crime_night['area_sq_km'].mean() if 'area_sq_km' in crime_night.columns else 'N/A',
    crime_night['population_density'].mean() if 'population_density' in crime_night.columns else 'N/A',
    crime_night['income'].mean() if 'income' in crime_night.columns else 'N/A',
    crime_night['crime_count'].mean() if 'crime_count' in crime_night.columns else 'N/A',
    crime_night['ntl_mean'].mean() if 'ntl_mean' in crime_night.columns else 'N/A',
    crime_night[crime_night['is_downtown'] == 1].shape[0] if 'is_downtown' in crime_night.columns else 'N/A'
]

# Add metrics to the comparison dataframe
comparison_metrics['Day'] = day_metrics
comparison_metrics['Night'] = night_metrics

# Display the comparison table
print("\nComparison of key metrics between day and night crime data:")
display(comparison_metrics)

# Create more detailed visualizations for the comparison
plt.figure(figsize=(16, 12))

# 1. Neighborhood comparison - bar chart of top neighborhoods by time category
if 'neighborhood' in crime_day.columns and 'neighborhood' in crime_night.columns:
    plt.subplot(2, 2, 1)
    
    # Get top 5 neighborhoods for each time category
    day_neighborhoods = crime_day['neighborhood'].value_counts().head(5)
    night_neighborhoods = crime_night['neighborhood'].value_counts().head(5)
    
    # Combine and plot
    comparison_df = pd.DataFrame({
        'Day': day_neighborhoods, 
        'Night': night_neighborhoods
    }).fillna(0)
    
    comparison_df.plot(kind='bar', ax=plt.gca())
    plt.title('Top 5 Neighborhoods: Day vs Night')
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')

# 2. Population comparison if available
if 'population' in crime_day.columns and 'population' in crime_night.columns:
    plt.subplot(2, 2, 2)
    
    # Create boxplots for population comparison
    pop_day = crime_day['population'].dropna()
    pop_night = crime_night['population'].dropna()
    
    plt.boxplot([pop_day, pop_night], labels=['Day', 'Night'])
    plt.title('Population Comparison')
    plt.ylabel('Population')

# 3. Income comparison if available
if 'income' in crime_day.columns and 'income' in crime_night.columns:
    plt.subplot(2, 2, 3)
    
    # Create boxplots for income comparison
    inc_day = crime_day['income'].dropna()
    inc_night = crime_night['income'].dropna()
    
    plt.boxplot([inc_day, inc_night], labels=['Day', 'Night'])
    plt.title('Income Comparison')
    plt.ylabel('Income')

# 4. Downtown vs non-downtown comparison
if 'is_downtown' in crime_day.columns and 'is_downtown' in crime_night.columns:
    plt.subplot(2, 2, 4)
    
    # Count downtown vs non-downtown crimes for day and night
    day_downtown = crime_day['is_downtown'].value_counts()
    night_downtown = crime_night['is_downtown'].value_counts()
    
    # Create comparison dataframe
    downtown_comp = pd.DataFrame({
        'Day': [day_downtown.get(1, 0), day_downtown.get(0, 0)],
        'Night': [night_downtown.get(1, 0), night_downtown.get(0, 0)]
    }, index=['Downtown', 'Non-Downtown'])
    
    downtown_comp.plot(kind='bar', stacked=True, ax=plt.gca())
    plt.title('Downtown vs Non-Downtown: Day vs Night')
    plt.ylabel('Count')
    plt.xticks(rotation=0)
    
plt.tight_layout()
plt.show()

# Calculate percentages for is_downtown
if 'is_downtown' in crime_day.columns and 'is_downtown' in crime_night.columns:
    day_downtown_pct = (crime_day['is_downtown'] == 1).mean() * 100
    night_downtown_pct = (crime_night['is_downtown'] == 1).mean() * 100
    
    print(f"\nPercentage of crimes in downtown areas:")
    print(f"Day: {day_downtown_pct:.2f}%")
    print(f"Night: {night_downtown_pct:.2f}%")
    
    # Statistical test for downtown proportion
    from scipy.stats import chi2_contingency
    
    # Create contingency table
    contingency = pd.crosstab(
        index=[crime_day['is_downtown'].tolist() + crime_night['is_downtown'].tolist()],
        columns=[['day'] * len(crime_day) + ['night'] * len(crime_night)]
    )
    
    # Perform chi-square test
    chi2, p, dof, expected = chi2_contingency(contingency)
    print(f"\nChi-square test for downtown proportion difference: p-value = {p:.4f}")
    print(f"Is the difference in downtown proportion statistically significant? {'Yes' if p < 0.05 else 'No'} (α=0.05)")
    
# Additional relevant information - Display neighborhoods that appear in one category but not the other
if 'neighborhood' in crime_day.columns and 'neighborhood' in crime_night.columns:
    day_neighborhoods = set(crime_day['neighborhood'].unique())
    night_neighborhoods = set(crime_night['neighborhood'].unique())
    
    only_day = day_neighborhoods - night_neighborhoods
    only_night = night_neighborhoods - day_neighborhoods
    
    print(f"\nNeighborhoods that appear only in day data: {len(only_day)}")
    if len(only_day) > 0:
        print(only_day)
    
    print(f"\nNeighborhoods that appear only in night data: {len(only_night)}")
    if len(only_night) > 0:
        print(only_night)
```

```python
# Visualize the day vs night comparison table
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Create a visualization of the comparison metrics
plt.figure(figsize=(14, 10))

# Get numeric metrics only for visualization
numeric_metrics = []
day_values = []
night_values = []

for i, metric in enumerate(comparison_metrics['Metric']):
    # Check if both day and night values are numeric
    try:
        day_val = float(comparison_metrics['Day'][i])
        night_val = float(comparison_metrics['Night'][i])
        numeric_metrics.append(metric)
        day_values.append(day_val)
        night_values.append(night_val)
    except (ValueError, TypeError):
        # Skip non-numeric values
        pass

# Create a DataFrame for the numeric metrics
numeric_df = pd.DataFrame({
    'Metric': numeric_metrics,
    'Day': day_values,
    'Night': night_values
})

# Calculate the percentage difference
numeric_df['Pct_Diff'] = ((numeric_df['Night'] - numeric_df['Day']) / numeric_df['Day'] * 100).round(2)

# Create a horizontal bar chart comparing day vs night for the numeric metrics
plt.subplot(2, 1, 1)
bar_width = 0.35
y_pos = np.arange(len(numeric_metrics))

plt.barh(y_pos - bar_width/2, numeric_df['Day'], bar_width, label='Day', color='skyblue')
plt.barh(y_pos + bar_width/2, numeric_df['Night'], bar_width, label='Night', color='navy')

# Add metric names as y-axis labels
plt.yticks(y_pos, numeric_metrics)
plt.xlabel('Value')
plt.title('Day vs Night Comparison of Key Metrics')
plt.legend()

# Normalize the values for comparison (if they have very different scales)
plt.subplot(2, 1, 2)
# Create a normalized version where for each metric, day = 100%
normalized_df = numeric_df.copy()
normalized_df['Night_Normalized'] = (normalized_df['Night'] / normalized_df['Day'] * 100).round(2)
normalized_df['Day_Normalized'] = 100  # Day is the baseline at 100%

# Sort by the percentage difference
normalized_df = normalized_df.sort_values('Pct_Diff')

plt.barh(normalized_df['Metric'], normalized_df['Pct_Diff'], color=['red' if x < 0 else 'green' for x in normalized_df['Pct_Diff']])
plt.axvline(x=0, color='black', linestyle='-', linewidth=0.5)
plt.title('Percentage Difference: Night vs Day (Day as baseline)')
plt.xlabel('Percentage Difference (%)')
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Add percentage values as text
for i, val in enumerate(normalized_df['Pct_Diff']):
    plt.text(val + (5 if val >= 0 else -5), 
             i, 
             f"{val}%", 
             va='center',
             ha='right' if val < 0 else 'left',
             fontweight='bold')

plt.tight_layout()
plt.show()

# Create a heatmap visualization of the complete comparison
try:
    # Create a wider comparison table for visualization
    vis_data = pd.DataFrame(columns=['Category', 'Day', 'Night', 'Difference', 'Pct_Difference'])
    
    for i, metric in enumerate(comparison_metrics['Metric']):
        day_val = comparison_metrics['Day'][i]
        night_val = comparison_metrics['Night'][i]
        
        try:
            day_val = float(day_val)
            night_val = float(night_val)
            diff = night_val - day_val
            pct_diff = (diff / day_val * 100) if day_val != 0 else float('inf')
            
            vis_data.loc[len(vis_data)] = [
                metric, day_val, night_val, diff, pct_diff
            ]
        except (ValueError, TypeError):
            # For non-numeric values, just use strings and None for difference
            vis_data.loc[len(vis_data)] = [
                metric, str(day_val), str(night_val), None, None
            ]
            
    # Display the enhanced comparison table
    print("\nEnhanced comparison table (with differences):")
    display(vis_data)
    
    # Create a heatmap for the numeric metrics
    numeric_metrics_data = vis_data.dropna(subset=['Difference', 'Pct_Difference'])
    
    if len(numeric_metrics_data) > 0:
        plt.figure(figsize=(12, len(numeric_metrics_data) * 0.8))
        
        # Normalize the percentage differences for better visualization
        norm_pct_diffs = numeric_metrics_data['Pct_Difference'].values.reshape(-1, 1)
        
        # Create a heatmap
        sns.heatmap(norm_pct_diffs, 
                    cmap='RdBu_r',
                    center=0,
                    annot=True, 
                    fmt=".1f",
                    yticklabels=numeric_metrics_data['Category'],
                    xticklabels=['Night vs Day (%)'],
                    cbar_kws={'label': 'Percentage Difference'})
        
        plt.title('Percentage Difference Heatmap: Night vs Day')
        plt.tight_layout()
        plt.show()

except Exception as e:
    print(f"Error creating heatmap visualization: {e}")
    
# Create a radar chart for comparing day and night metrics
try:
    if len(numeric_df) >= 3:  # Need at least 3 metrics for a radar chart
        plt.figure(figsize=(10, 8))
        
        # Normalize values for radar chart (scale from 0 to 1)
        radar_df = numeric_df.copy()
        max_vals = np.maximum(radar_df['Day'], radar_df['Night'])
        radar_df['Day_Scaled'] = radar_df['Day'] / max_vals
        radar_df['Night_Scaled'] = radar_df['Night'] / max_vals
        
        # Get the number of variables
        categories = radar_df['Metric']
        N = len(categories)
        
        # Create angles for each metric (divide the plot into equal parts)
        angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()
        angles += angles[:1]  # Close the loop
        
        # Add the day and night values, and close the loop
        day_values = radar_df['Day_Scaled'].tolist()
        day_values += day_values[:1]
        night_values = radar_df['Night_Scaled'].tolist()
        night_values += night_values[:1]
        
        # Set up the plot
        ax = plt.subplot(111, polar=True)
        
        # Plot day and night
        ax.plot(angles, day_values, 'o-', linewidth=2, label='Day', color='skyblue')
        ax.fill(angles, day_values, alpha=0.25, color='skyblue')
        ax.plot(angles, night_values, 'o-', linewidth=2, label='Night', color='navy')
        ax.fill(angles, night_values, alpha=0.25, color='navy')
        
        # Set category labels
        plt.xticks(angles[:-1], categories)
        
        # Add legend
        plt.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1))
        plt.title('Radar Chart: Day vs Night Crime Characteristics', size=15)
        
        plt.tight_layout()
        plt.show()
except Exception as e:
    print(f"Error creating radar chart: {e}")
```
